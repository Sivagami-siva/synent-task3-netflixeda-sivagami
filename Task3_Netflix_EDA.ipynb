```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("netflix_titles.csv")
print(data.head())
print(data.info())
#Check Missing Values
print(data.isnull().sum())
print(data.describe())
#Count Movies vs TV Shows
type_count = data['type'].value_counts()
type_count.plot(kind='bar',color='purple')
plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
#Top 10 Countries
country_count = data['country'].value_counts().head(10)
country_count.plot(kind='bar',color='red')
plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
#Release Year Trend
year_count = data['release_year'].value_counts().sort_index()
year_count.plot()
plt.title("Netflix Content Release Trend")
plt.xlabel("Year")
plt.ylabel("Number of Releases")
plt.show()
#Top Ratings
rating_count = data['rating'].value_counts().head(10)
rating_count.plot(kind='bar')
plt.title("Top Ratings")
plt.xlabel("Ratings")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
#Correlation Analysis
correlation = data.corr(numeric_only=True)
print(correlation)
```

1. Netflix contains more Movies than TV Shows.
2. United States produces the highest content.
3. Netflix content increased rapidly after 2015.
4. TV-MA is one of the most common ratings.
